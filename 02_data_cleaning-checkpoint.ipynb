```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

print("Libraries imported successfully!")
```

```python
raw_file = "../data/raw/BK_AUG2026.xlsx"
processed_dir = "../data/processed"

os.makedirs(processed_dir, exist_ok=True)

print("Raw file:", os.path.abspath(raw_file))
print("Processed folder:", os.path.abspath(processed_dir))
```

```python
if os.path.exists(raw_file):
    print("✅ Raw Excel file found")
else:
    print("❌ Raw Excel file not found")
```

```python
excel_file = pd.ExcelFile(raw_file)

print("Available sheets:")
print(excel_file.sheet_names)
```

```python
dataframes = []

for sheet in excel_file.sheet_names:
    temp_df = pd.read_excel(
        raw_file,
        sheet_name=sheet
    )
    
    temp_df["source_sheet"] = sheet
    
    dataframes.append(temp_df)

df = pd.concat(
    dataframes,
    ignore_index=True
)

print("Combined dataset shape:", df.shape)
```

```python
print("Original column names:")

for column in df.columns:
    print(column)
```

```python
print("Original column names:")

for column in df.columns:
    print(column)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print("Standardized columns:")
print(df.columns.tolist())
```

```python
df = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df))
```

```python
if "BKDATE" in df.columns:
    print("✅ BKDATE column found")
else:
    print("❌ BKDATE column not found")
```

```python
df["BKDATE"] = pd.to_datetime(
    df["BKDATE"],
    errors="coerce"
)
```

```python
invalid_dates = df["BKDATE"].isna().sum()

print("Invalid/missing booking dates:", invalid_dates)
```

```python
df = df.dropna(
    subset=["BKDATE"]
).copy()

print("Rows after removing invalid dates:", len(df))
```

```python
print("First booking date:", df["BKDATE"].min())
print("Last booking date :", df["BKDATE"].max())
```
